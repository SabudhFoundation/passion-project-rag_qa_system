```python
from docling.chunking import HybridChunker
from pathlib import Path
import os
from docling_core.transforms.chunker.tokenizer.huggingface import (
    HuggingFaceTokenizer,
)
from transformers import AutoTokenizer
from docling_core.transforms.chunker.hierarchical_chunker import (
    ChunkingDocSerializer,
    ChunkingSerializerProvider,
)
from docling_core.transforms.serializer.markdown import (
    MarkdownParams,
    MarkdownTableSerializer,
)
from docling.document_converter import DocumentConverter
print(HuggingFaceTokenizer.model_fields)
csv_result = DocumentConverter().convert(source=r"E:\ElectricCarDataset\ElectricCarData_Clean.csv")
csv_doc = csv_result.document
output_file=Path("table_output.txt")
with open(output_file, "w", encoding="utf-8") as f:
    #f.write(csv_doc)
    #f.write("\n\n")
    #f.write(csv_doc.tables)
    f.write("\n\n")
    for item in csv_doc.tables:
        cells_list=item.data
        break
    #f.write(list[cells_list])
    f.write("\n\n\n\n\n")

    for cell in cells_list.table_cells:
        f.write(str(cell)+"\n\n")
        
os.system(f'notepad "{output_file}"')

class MDTableSerializerProvider(ChunkingSerializerProvider):
    def get_serializer(self, doc) -> ChunkingDocSerializer:
        return ChunkingDocSerializer(
            doc=doc,
            table_serializer=MarkdownTableSerializer(),
        )
tokenizer = HuggingFaceTokenizer(
    tokenizer=AutoTokenizer.from_pretrained("jinaai/jina-embeddings-v3"),
    max_tokens=512
)
csv_chunker = HybridChunker(
    tokenizer=tokenizer,
    repeat_table_header=True,
    serializer_provider=MDTableSerializerProvider()
)

chunks = list(csv_chunker.chunk(csv_doc))

print(f"Total chunks: {len(chunks)}")
print(chunks)

"""for i, chunk in enumerate(chunks):
    print(f"\n{'='*50}")
    print(f"CHUNK {i}")
    print(f"{'='*50}")
    print(f"type(chunk): {type(chunk)}")
    print(f"chunk.meta: {chunk.meta}")
    print(f"\nchunk.text:\n{chunk.text}")"""
    
```

```python
import json
file_path=r"C:\Users\DELL\Downloads\hotpot_train_v1.1.json"
if file_path:
    with open(file_path, 'r') as f:
        data = json.load(f)
        
```

```python
import uuid
table1=[] #will store question along with _id,followed by supporting facts 
table2=[] #will store context used to answer these questions

for index,json_dict in enumerate(data):
    if not json_dict:
        continue
    #---building table 2 ---
    context=json_dict["context"]
    seen_labels=set()
    label_dict={}
    for context_data in context:
        label,ctx_data=context_data[0],context_data[1]
        if label not in seen_labels:
            seen_labels.add(label)  
            table_data={
                "_id":str(uuid.uuid4()),
                "label":label,
                "context":ctx_data
            }    
            table2.append(table_data)  
            label_dict[label]=table_data["_id"]
            
    #---Build table 1---
    question=json_dict["question"]
    id=json_dict["_id"]
    supporting_facts=json_dict["supporting_facts"]
    supporting_facts_list = []
    for sf in supporting_facts:
        label,sf_index=sf[0],sf[1]
        if label in label_dict:
            supporting_facts_list.append({
                "label_id": label_dict[label],
                "sentence_index": sf_index  # ✅ renamed: it's a sentence index, not context index
            })
    table1.append({
        "_id": id,
        "question": question,
        "supporting_facts": supporting_facts_list,
    })
```
